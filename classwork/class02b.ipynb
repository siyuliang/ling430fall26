#### activity: the turkish locative

```python
# print the locative form of every word: -da / -de / -ta / -te
# 1. find the last vowel of the word
# 2. back vowel -> a, front vowel -> e
# 3. last letter voiceless -> t, otherwise d
# expected output, one per line:
#   evde okulda kebapta yoğurtta baklavada pilavda dönerde
#   çayda kahvede otobüste parkta otelde sokakta denizde
words = ["ev", "okul", "kebap", "yoğurt", "baklava", "pilav", "döner", "çay", "kahve", "otobüs", "park", "otel", "sokak", "deniz"]
vowels = "aeıioöuü"
back = "aıou"
voiceless = "çfhkpsşt"
for w in words:
    last_vowel = ""
    for ch in w:
        if ch in vowels:
            last_vowel = ch
    print(w, last_vowel)
```

```python
# the turkish locative
words = ["ev", "okul", "kebap", "yoğurt", "baklava", "pilav", "döner", "çay", "kahve", "otobüs", "park", "otel", "sokak", "deniz"]
vowels = "aeıioöuü"
back = "aıou"
voiceless = "çfhkpsşt"
for w in words:
    last_vowel = ""
    for ch in w:
        if ch in vowels:
            last_vowel = ch
    if last_vowel in back:
        if w[-1] in voiceless:
            print(w + "ta")
        else:
            print(w + "da")
    else:
        if w[-1] in voiceless:
            print(w + "te")
        else:
            print(w + "de")
```

```python
# extra challenge
for w in words:
    last_vowel = ""
    for ch in w:
        if ch in vowels:
            last_vowel = ch
    if last_vowel in back:
        print(w + "larda")
    else:
        print(w + "lerde")
```
