#### getting a corpus

```python
# the complete works of Shakespeare (Project Gutenberg #100)
from urllib.request import urlopen

raw = urlopen("https://www.gutenberg.org/cache/epub/100/pg100.txt").read().decode("utf-8-sig")
start = raw.find("*** START")
start = raw.find("\n", start) + 1
end = raw.find("*** END")
text = raw[start:end].strip()
len(text)
```

```python
print(text[:1000])
```

```python
words = text.split()
len(words)
```

```python
len(set(words))
```

```python
words.count("love")
```

```python
# first fix: capitalization
lowered = text.lower().split()
len(set(lowered))
```

```python
# second fix: punctuation
# no need to understand it for now; we will learn all about this later in the class
import re
cleaned_words = [re.sub(r"^[\W_]+|[\W_]+$", "", w) for w in text.lower().split()]
cleaned_words= [w for w in cleaned_words if w]
len(set(cleaned_words))
```

```python
cleaned_words.count("love")
```

```python
# the frequency distribution
import matplotlib.pyplot as plt
from collections import Counter

counts = Counter(cleaned_words)
freqs = sorted(counts.values(), reverse=True)
ranks = range(1, len(freqs) + 1)

plt.figure(figsize=(9, 3.5))
plt.plot(ranks, freqs)
plt.xlabel("rank")
plt.ylabel("frequency")
```

```python
# no log scale, for reference: the head, with the words showing
items = [(w, n) for w, n in counts.most_common() if not w.isdigit()]   # skip the edition's line numbers
head = items[:20]

plt.figure(figsize=(9, 3.5))
plt.bar([w for w, n in head], [n for w, n in head])
plt.title("head: the 20 most frequent")
plt.ylabel("frequency")
plt.show()
```

```python
# and the tail
tail = items[-20:]

plt.figure(figsize=(30, 5))
plt.bar([w for w, n in tail], [n for w, n in tail])
plt.title("tail: 20 of the rarest")
plt.ylabel("frequency")
plt.show()
```

#### Zipf's law

```python
counts["the"] / len(cleaned_words)
```

```python
# percentage of 100 most frequent words
sum(n for w, n in counts.most_common(100)) / len(cleaned_words)
```

```python
# words used exactly once
sum(1 for n in counts.values() if n == 1)
```

```python
# zipf's law: frequency vs rank, both axes log
plt.loglog(ranks, freqs)
plt.loglog(ranks, [freqs[0] / r for r in ranks], "--")   # a perfect 1/rank line, for comparison
plt.xlabel("rank")
plt.ylabel("frequency")
```

#### the class corpus

```python
import pandas as pd

sheet = "PASTE_CUSTOM_LINK"
sheet = sheet.split("/edit")[0] + "/export?format=csv&gid=" + sheet.split("gid=")[-1] 
pot = " ".join(pd.read_csv(sheet).iloc[:, 1].dropna())
len(pot)
```

```python
class_words = [re.sub(r"^[\W_]+|[\W_]+$", "", w) for w in pot.lower().split()]
class_words = [w for w in class_words if w]
class_counts = Counter(class_words)
print(len(class_words), len(class_counts))
```

```python
class_counts.most_common(10)
```

```python
for c in [counts, class_counts]:
    f = sorted(c.values(), reverse=True)
    plt.loglog(range(1, len(f) + 1), f)
plt.xlabel("rank")
plt.ylabel("frequency")
plt.legend(["shakespeare", "us"])
```

```python
# what share of our words appeared exactly once?
sum(1 for n in class_counts.values() if n == 1) / len(class_counts)
```
