```python
# 1. the turkish locative
words = ["ev", "okul", "kebap", "yoğurt", "baklava", "pilav", "döner", "çay", "kahve", "otobüs", "park", "otel", "sokak", "deniz"]
vowels = "aeıioöuü"
back = "aıou"
voiceless = "çfhkpsşt"

def locative(word):
    last_vowel = ""
    for ch in word:
        if ch in vowels:
            last_vowel = ch
    if last_vowel in back:
        if word[-1] in voiceless:
            return word + "ta"
        return word + "da"
    if word[-1] in voiceless:
        return word + "te"
    return word + "de"

for w in words:
    print(locative(w))
```

```python
# extra challenge
def plural(word):
    last_vowel = ""
    for ch in word:
        if ch in vowels:
            last_vowel = ch
    if last_vowel in back:
        return word + "lar"
    return word + "ler"

for w in words:
    print(locative(plural(w)))
```
