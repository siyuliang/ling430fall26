```python
raw = "strange women lying in ponds distributing"
porter = nltk.PorterStemmer()
[porter.stem(w) for w in raw.split()]
```

```python
lancaster = nltk.LancasterStemmer()
for w in ["lying", "women", "ceremony", "distributing", "power"]:
    print(w, porter.stem(w), lancaster.stem(w))
```
