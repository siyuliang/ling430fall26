#### setup

```python
import nltk
from math import log
nltk.download("gutenberg", quiet=True)
from nltk.corpus import gutenberg

tokens = list(gutenberg.words("austen-emma.txt"))
len(tokens)
```

#### probability

a probability is a fraction: the cases you care about over all the cases. 

in week 2 we counted with a dictionary; 

nltk's FreqDist does the same in one line, and `.freq()` is the fraction.

```python
# the brute-force way with counts
counts = {}
for w in tokens:
    counts[w] = counts.get(w, 0) + 1
counts["the"], len(tokens)
```

```python
fd = nltk.FreqDist(tokens)
fd["the"], fd.N()
```

```python
fd.most_common(3)
```

```python
fd.freq("the")
```

```python
# the probabilities of all the outcomes add up to 1
sum(fd.freq(w) for w in fd)
```

conjunction: both things at once. the pair (so, well) is a case where the token is *so* and the next token is *well*.

```python
bigrams = nltk.FreqDist(nltk.bigrams(tokens))
bigrams[("so", "well")]
```

```python
bigrams.freq(("so", "well"))
```

#### conditional probability


```python
# P(well | so): of the tokens that are so, how many are followed by well?
bigrams[("so", "well")] / fd["so"]
```

```python
# compared with P(well) over the whole novel
fd.freq("well")
```

```python
cfd = nltk.ConditionalFreqDist(nltk.bigrams(tokens))
cfd["so"].freq("well")
```

```python
cfd["so"].most_common(5)
```

conditional probabilities are not commutative: the same 173 pairs can have two different denominators and results

```python
fd["Miss"], fd["Woodhouse"], bigrams[("Miss", "Woodhouse")]
```

```python
# P(Woodhouse | after Miss) and P(Miss | before Woodhouse)
bigrams[("Miss", "Woodhouse")] / fd["Miss"], bigrams[("Miss", "Woodhouse")] / fd["Woodhouse"]
```

#### Bayes’ theorem

P(A | B) = P(A) × P(B | A) / P(B)

```python
prior = 0.01
likelihood = 0.8
false_positive = 0.096
p_positive = prior * likelihood + (1 - prior) * false_positive
p_positive
```

```python
prior * likelihood / p_positive
```

```python
p_miss, p_wood = fd.freq("Miss"), fd.freq("Woodhouse")
p_wood_given_miss = cfd["Miss"].freq("Woodhouse")
p_miss * p_wood_given_miss / p_wood
```

#### estimating probabilities

maximum likelihood: the parameter value that makes the observed data most probable. for counts, it is always the relative frequency.

```python
# the coin came up H H H T T. which value of theta makes that most probable?
for t in [0.4, 0.5, 0.6, 0.7]:
    print(t, round(t**3 * (1 - t)**2, 5))
```

```python
# a finer grid
best_t = 0
best_p = 0
for i in range(1, 100):
    t = i / 100
    p = t**3 * (1 - t)**2
    if p > best_p:
        best_t = t
        best_p = p
best_t, best_p
```

small numbers: products of probabilities vanish; sums of log probabilities do not.

```python
0.1 ** 100, 0.1 ** 1000
```

```python
sum(log(0.1) for i in range(1000))
```
