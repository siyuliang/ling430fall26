#### setup

```python
import random
import nltk
from nltk.corpus import movie_reviews
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix
nltk.download("movie_reviews", quiet=True)
```

#### training

```python
ids = list(movie_reviews.fileids())
random.seed(430)
random.shuffle(ids)
texts = [movie_reviews.raw(f) for f in ids]
labels = [movie_reviews.categories(f)[0] for f in ids]
train_texts, test_texts = texts[:1600], texts[1600:]
train_labels, test_labels = labels[:1600], labels[1600:]

vectorizer = CountVectorizer()
model = MultinomialNB()
model.fit(vectorizer.fit_transform(train_texts), train_labels)
```

#### testing

```python
pred = model.predict(vectorizer.transform(test_texts))
accuracy_score(test_labels, pred)
```

#### confusion matrix

```python
confusion_matrix(test_labels, pred, labels=["pos", "neg"])
```

```python
def show(gold, predicted, n=3):
    for text, g, p in zip(test_texts, test_labels, pred):
        if g == gold and p == predicted:
            print(text[:400])
            print("---")
            n = n - 1
            if n == 0:
                break
```
