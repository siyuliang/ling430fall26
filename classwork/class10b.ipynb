#### setup

```python
from nltk import CFG, ChartParser

L1 = CFG.fromstring("""
S -> NP VP | Aux NP VP | VP
NP -> Pronoun | Proper-Noun | Det Nominal
Nominal -> Noun | Nominal Noun | Nominal PP
VP -> Verb | Verb NP | Verb NP PP | Verb PP | VP PP
PP -> Preposition NP
Det -> 'that' | 'this' | 'the' | 'a'
Noun -> 'book' | 'flight' | 'meal' | 'money'
Verb -> 'book' | 'include' | 'prefer'
Pronoun -> 'I' | 'she' | 'me'
Proper-Noun -> 'Houston' | 'United'
Aux -> 'does'
Preposition -> 'from' | 'to' | 'on' | 'near' | 'through'
""")
```

#### parsing with nltk

```python
parser = ChartParser(L1)
sent = "book the flight through Houston".split()
trees = list(parser.parse(sent))
len(trees)
```

```python
for tree in trees:
    print(tree)
```

#### activity 1

part 2: parse the sentence with the grammar from the slide. how many trees do you get, and which one comes first? then add a word to the lexicon and parse a sentence of your own

```python
grammar = CFG.fromstring("""
S -> NP VP
NP -> Pronoun | Det Nom
Nom -> Noun | Nom PP
VP -> Verb NP | VP PP
PP -> Prep NP
Pronoun -> 'I'
Verb -> 'shot'
Det -> 'an' | 'my'
Noun -> 'elephant' | 'pajamas'
Prep -> 'in'
""")

parser = ChartParser(grammar)
sent = "I shot an elephant in my pajamas".split()
trees = list(parser.parse(sent))
print(len(trees))
for tree in trees:
    print(tree)
```
