#### fastText multilingual examples

```python
# fastText (Bojanowski et al. 2017) trained one space per Wikipedia, 294 languages: https://fasttext.cc/docs/en/pretrained-vectors.html
# the files are big (1 to 6 GB), but they are sorted by frequency, so we stream only the n most frequent words
import urllib.request, itertools, os
from gensim.models import KeyedVectors

def load_fasttext(lang, n=50000):
    path = f"wiki.{lang}.{n}.vec"
    if not os.path.exists(path):
        url = f"https://dl.fbaipublicfiles.com/fasttext/vectors-wiki/wiki.{lang}.vec"
        with urllib.request.urlopen(url) as r, open(path, "w") as f:
            dims = r.readline().decode().split()[1]
            f.write(f"{n} {dims}\n")
            for line in itertools.islice(r, n):
                f.write(line.decode())
    return KeyedVectors.load_word2vec_format(path)

hi = load_fasttext("hi")   # Hindi, about 130 MB
len(hi)
```

```python
# a Hindi grammar says: masculine -aa -> feminine -ii, as in लड़का laRkaa "boy" -> लड़की laRkii "girl"
# लड़का : लड़की :: बेटा (beTaa, son) : ?
hi.most_similar(positive=["लड़की", "बेटा"], negative=["लड़का"], topn=3)
```

```python
# the same for धोबी (dhobii, washerman); the feminine धोबिन is too rare to be in the vocabulary, so what do we get?
hi.most_similar(positive=["लड़की", "धोबी"], negative=["लड़का"], topn=3)
```
