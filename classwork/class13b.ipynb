#### Whisper

```python
!pip install -q openai-whisper
import whisper
model = whisper.load_model("base")
```

#### a recording

```python
# download the recording from Canvas, then choose it here
from google.colab import files
from IPython.display import Audio
audio = list(files.upload())[0]
Audio(audio)
```

#### transcribing

```python
# a warning about FP16 is fine
result = model.transcribe(audio)
result["text"]
```

#### language

```python
# Whisper guesses the language from the first 30 seconds
result["language"]
```

```python
# we can also tell it the language, with a code such as "en", "es", "zh", "vi" or "ko"
model.transcribe(audio, language="en")["text"]
```

```python
# and we can ask for an English translation instead of a transcript
model.transcribe(audio, task="translate")["text"]
```
